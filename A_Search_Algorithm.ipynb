```python
import heapq

goal_state = (1,2,3,8,0,4,7,6,5)

moves = {
    'U': -3,
    'D': 3,
    'L': -1,
    'R': 1
}

def is_valid_move(blank_index, move):
    if move == 'L' and blank_index % 3 == 0:
        return False
    if move == 'R' and blank_index % 3 == 2:
        return False
    if move == 'U' and blank_index < 3:
        return False
    if move == 'D' and blank_index > 5:
        return False
    return True

def apply_move(state, move):
    blank_index = state.index(0)
    new_blank = blank_index + moves[move]
    new_state = list(state)
    new_state[blank_index], new_state[new_blank] = new_state[new_blank], new_state[blank_index]
    return tuple(new_state)

def misplaced_tiles(state):
    return sum(1 for i in range(9) if state[i] != 0 and state[i] != goal_state[i])

def manhattan_distance(state):
    dist = 0
    for i in range(9):
        if state[i] != 0:
            goal_pos = goal_state.index(state[i])
            dist += abs(i//3 - goal_pos//3) + abs(i%3 - goal_pos%3)
    return dist


def a_star(start_state, heuristic):
    frontier = []
    heapq.heappush(frontier, (0, start_state, [start_state]))
    explored = set()

    while frontier:
        f, state, path = heapq.heappop(frontier)

        if state == goal_state:
            return path

        if state in explored:
            continue
        explored.add(state)

        g = len(path) - 1
        h = heuristic(state)

        blank_index = state.index(0)
        for move in moves:
            if is_valid_move(blank_index, move):
                new_state = apply_move(state, move)
                if new_state not in explored:
                    new_path = path + [new_state]
                    new_g = g + 1
                    new_h = heuristic(new_state)
                    new_f = new_g + new_h
                    heapq.heappush(frontier, (new_f, new_state, new_path))
    return None

start_state = (2,8,3,
               1,6,4,
               7,0,5)

print("Using Misplaced Tiles Heuristic:")
solution1 = a_star(start_state, misplaced_tiles)
if solution1:
    print("Solution found in", len(solution1)-1, "moves!")
    for step in solution1:
        print(step)

print("\nUsing Manhattan Distance Heuristic:")
solution2 = a_star(start_state, manhattan_distance)
if solution2:
    print("Solution found in", len(solution2)-1, "moves!")
    for step in solution2:
        print(step)
```
