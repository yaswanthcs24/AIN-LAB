```python
from collections import deque
goal_state = (1,2,3,4,5,6,7,8,0)
moves = {
    'U': -3,
    'D': 3,
    'L': -1,
    'R': 1
}
def is_valid_move(blank_index, move):
    if move == 'L' and blank_index % 3 == 0:
        return False
    if move == 'R' and blank_index % 3 == 2:
        return False
    if move == 'U' and blank_index < 3:
        return False
    if move == 'D' and blank_index > 5:
        return False
    return True
def apply_move(state, move):
    blank_index = state.index(0)
    new_blank = blank_index + moves[move]
    new_state = list(state)
    new_state[blank_index], new_state[new_blank] = new_state[new_blank], new_state[blank_index]
    return tuple(new_state)
def dls(state, depth, path):
    if state == goal_state:
        return path
    if depth == 0:
        return None
    blank_index = state.index(0)
    for move in moves:
        if is_valid_move(blank_index, move):
            new_state = apply_move(state, move)
            if new_state not in path:  # avoid cycles
                result = dls(new_state, depth-1, path+[new_state])
                if result:
                    return result
    return None

def ids(start_state, max_depth):
    for depth in range(max_depth+1):
        result = dls(start_state, depth, [start_state])
        if result:
            return result
    return None

# Example start state
start_state = (1,2,3,
               4,5,6,
               0,7,8)

solution = ids(start_state, 20)

if solution:
    print("Solution found in", len(solution)-1, "moves!")
    for step in solution:
        print(step)
else:
    print("No solution within depth limit.")
print("KATASANI YASWANTH REDDY-1WN24CS133")
```
